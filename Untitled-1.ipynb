```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import fsolve
```

```python
def s(x):
    return -1.6*x**2+1.3*x+141.3
def r(x):
    return 1.4*x**2-5.6*x+12.6

x = np.arange(-10, 10, 0.1)
plt.plot(x, s(x), label = 's(x)')
plt.plot(x, r(x), label = 'r(x)')
plt.title('Nombre del gráfico')
plt.ylabel('Nombre eje y')
plt.xlabel('Nombre eje x')
plt.legend()
plt.grid()
plt.show()
```

```python
def sol(x):
    return s(x)-r(x)
xo = np.linspace(-10, 10,3) #'cantidad de valores del arreglo', se refiere a indicar el número de soluciones que tiene la ecuación
solucion = fsolve(sol, xo)
print(f"{solucion}")
```

```python
def D(t):
    return 50*np.exp(0.3*t)
def D1 (t):
    return D(t)-551
xo = np.linspace(0, 8, 1)
solucion = fsolve(D1, xo)
print(f"{solucion}")
```
